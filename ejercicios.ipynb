## Ejercicio 1
Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
texto = input("Escribe una frase: ")
dict_letras = {}

def contador_letras(texto):
    """
    Cuenta las letras que hay en un string

    Args:
        texto (_str): se genera a través de un input

    Returns:
        dict: la clave es el caracter presente en el string y el valor son el número de veces que aparece
    """
    for letra in texto:
        if letra != " ": # El ejercicio pedía que no se consideren los espacios
            if letra not in dict_letras:
                dict_letras[letra] = 1
            else: 
                dict_letras[letra] += 1
    
    return dict_letras

print(contador_letras(texto))

```

```python
lista_num1 = [85, 39, 24, 8, 3]
lista_num2 = [1, 4, 77, 3, 8]

def resta_listas (lista1, lista2):
    """
    Resta dos listas y devuelve la diferencia de cada posición
    Args:
        lista1 (lista): lista de números enteros o float, será el minuendo
        lista2 (lista): lista de números enteros o float, será el sustraendo 

    Returns:
       lista: lista de números enteros o float, cada posición es la diferencia de las listas
    """
    resultado = list(map(lambda x, y : x - y, lista1, lista2))
    return resultado

print(resta_listas(lista_num1, lista_num2))
```

## Ejercicio 5

Escribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado (por defecto 5). La función debe calcular la media de los números en la lista y determinar si la media es mayor o igual que nota_aprobado. Si es así, el estado será "aprobado"; de lo contrario, "suspenso". La función debe devolver una tupla que contenga la media y el estado.

```python
notas = [4, 6, 9, 1, 3, 7, 4]

def calcular_media (lista_num, nota_aprobado = 5):
    """
    Calcula la media de una lista de números y compara si está aprobado o suspenso
    Args:
        lista_num (lista): lista de números enteros o float
        nota_aprobado (int, optional): La nota con la que se compara el aprobado o suspenso, por defecto es 5

    Returns:
        tupla: devuelve una tupla donde el primer valor es la media y el segundo es el estado (aprobado, suspenso)
    """
    suma = 0
    for numero in lista_num:
        suma += numero
    media = round(suma/len(lista_num))

    if media >= nota_aprobado:
        return media, "Aprobado"
    else:
        return media, "Suspenso"

    
calcular_media(notas) 

```

## Ejercicio 30

Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero en diferente orden.

```python
def anagrama (palabra1, palabra2):
    """
    Comprueba si dos palabras son anagramas

    Args:
        palabra1 (str): primera palabra
        palabra2 (str): segunda palabra

    Returns:
        bool: la función retorna un booleano, también imprime por consola una frase en función de si son o no anagramas
    """
    if palabra1 == palabra2:
        print("Oye has introducido la misma palabra")
        return 
    else:
        set_palabra1 = set(palabra1)
        set_palabra2 = set(palabra2)
        if set_palabra1 == set_palabra2:
            print(f"{palabra1} y {palabra2} son anagramas")
            return True
        else:
            print(f"{palabra1} y {palabra2} no son anagramas")
            return False
    

anagrama("riesgo", "sergio")
```

Crea la clase Arbol

```python
class Arbol:
    def __init__(self):
        self.tronco = 1
        self.ramas = []

    def crecer_tronco(self):
        self.tronco += 1

    def nueva_rama (self):
        self.ramas.append(1)

    def crecer_ramas(self):
        for i in range(len(self.ramas)):
            self.ramas[i] += 1

    def quitar_rama (self, posicion):
        self.ramas.pop(posicion)
        

    def info_arbol (self):
        return f"La longitud del tronco es de {self.tronco} unidades, el número de ramas es {len(self.ramas)} y las longitudes de las ramas son {self.ramas}"

```

```python
# CASOS DE USO:


mi_arbol = Arbol()
## Hacer crecer el tronco una unidad
mi_arbol.crecer_tronco()


## Añadir una nueva rama
mi_arbol.nueva_rama()


## Hacer crecer todas las ramas una unidad
mi_arbol.crecer_ramas()


## Añadir dos nuevas ramas
for i in range(2):
    mi_arbol.nueva_rama()


## Retirar la rama situada en la posición 2
mi_arbol.quitar_rama(2) 


## Obtener información sobre el árbol
mi_arbol.info_arbol() 
```

Crea la clase UsuarioBanco

```python
class UsuarioBanco:
    def __init__(self, nombre, saldo, cuenta_corriente):
        self.nombre = nombre
        self.saldo = saldo
        self.cuenta_corriente = cuenta_corriente

    def retirar_dinero(self, cantidad):
        if cantidad > self.saldo:
            raise ValueError("¡No tienes suficiente dinero en la cuenta!")
        else:
            self.saldo -= cantidad

    def transferir_dinero(self, remitente, cantidad):
        if cantidad > remitente.saldo:
            raise ValueError("No tienes suficiente dinero en la cuenta!")
        else:
            remitente.saldo -= cantidad
            self.saldo += cantidad

    def agregar_dinero(self, cantidad):
        self.saldo += cantidad
```

```python
# CASOS DE USO

Alicia = UsuarioBanco("Alicia", 100, True)
Bob = UsuarioBanco("Bob", 50, True)


## Agregar 20 unidades al saldo de Bob
Bob.agregar_dinero(20)


## Transferir 80 unidades de Bob a Alicia
Alicia.transferir_dinero(Bob, 80)


## Retira 50 unidades del saldo de Alicia
Alicia.retirar_dinero(50)
```

## Ejercicio 36

Crea una función llamada procesar_texto

```python

def procesar_texto(texto, opcion, *args):
    if opcion == "contar":
        return contar_palabras(texto)
        
    elif opcion == "reemplazar":
        palabra_original = args[0]
        palabra_nueva = args[1]
        return reemplazar_palabras(texto, palabra_original, palabra_nueva)
        
    elif opcion == "eliminar":
        palabra_a_borrar = args[0]
        return eliminar_palabra(texto, palabra_a_borrar)

def contar_palabras(texto):
    palabras = texto.split()
    diccionario = {}
    
    for p in palabras:
        if p not in diccionario:
            diccionario[p] = 1
        else:
            diccionario[p] += 1
        
    return diccionario

def reemplazar_palabras(texto, palabra_original, palabra_nueva):
    texto = texto.replace(palabra_original, palabra_nueva)
    return texto


def eliminar_palabra(texto, palabra):
    texto = texto.replace(palabra, "")
    return texto

```

```python
# CASOS DE USO 

texto1 = "Lorem ipsum dolor sit amet, consectetur adipiscing elit"
print(procesar_texto(texto1, "contar"))

texto2 = "Fui a comer y a comer fui"
print(procesar_texto(texto2, "reemplazar", "comer", "bailar"))

texto3 = "A veces pienso cosas interesantes a veces no"
print(procesar_texto(texto3, "eliminar", "cosas interesantes"))
```
